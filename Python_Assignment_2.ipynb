## SETS

```python
my_set={'a','e','i','o','u','a','a','i'}
print(my_set)
```

Tried to create a set with 8 alphabets/values but the outcome contains only 5 alphabets/values.
The reason for this outcome is that the set doesn't favour duplicate values.
As the alphabets 'a' and 'i' are repeating more than one times,so it eliminated the repeated values.

```python
my_set[4]='s'
print(my_set)
```

One of the notable features of a set is that it is not ordered.That is, the values don't carry a specific order or index.
Therefore, indexing is not possible when it comes to sets.That is the reason why it throws an error as it doesn't support adding a new value to any specific position,including the last one. 

```python
set1={1, 3, 5, 7, 9}
set2={2, 3, 5, 8, 10}
print(set1)
print(set2)
```

```python
set1={1, 3, 5, 7, 9}
set2={2, 3, 5, 8, 10}
print(set1)
print(set2)
new_set=set1.union(set2)
print(new_set)
```

```python
set1={1, 3, 5, 7, 9}
set2={2, 3, 5, 8, 10}
print(set1)
print(set2)
common_num=set1.intersection(set2)
print(common_num)
```

## OPERATORS AND CONDITIONAL STATEMENTS
